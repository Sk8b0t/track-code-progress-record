1. Constant Window(for k=4)[max sum of 4 consecutive elements]

```python
arr=[-1,2,3,5,6,-3,5,2]
k=4
l=0
r=k-1
sum=0
max_sum=0
for i in range(k):
    sum+=arr[i]

while(r<len(arr)-1):
    sum-=arr[l]
    l+=1
    r+=1
    sum+=arr[r]
    max_sum=max(max_sum,sum)
print(max_sum)
```

```python
arr=[-1,2,3,5,6,-3,5,2]
# method -1 Brute Force
k=14
max_len=0
for i in range(len(arr)):
    sum=0
    for j in range(i,len(arr)):
        sum+=arr[j]
        if sum<=k:
            max_len=max(max_len,j-i+1)
print(max_len)
```

```python

k=14
l=r=ml=0
sum=0
while(r<len(arr)):
    sum+=arr[r]
    while(sum>k):
        sum-=arr[l]
        l+=1
    if sum<=k:
        ml=max(ml,r-l+1)
    r+=1
    
print(ml)
    
```

Maximum points you can obtain from cards

```python
arr=[1,7,4,5,6,3,5,2]
k=4
rs=ls=0

for i in range(k):
    ls+=arr[i]

ms=ls
rind=len(arr)-1

for i in range(k-1,-1,-1):
    ls-=arr[i]
    rs+=arr[rind]
    rind-=1
    ms=max(ms,ls+rs)
print(ms)
```
